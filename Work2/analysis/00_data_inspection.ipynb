# Inspect the datasets
Quick look at fold 0 of each dataset.


```python
import sys
from pathlib import Path
import pandas as pd

# Setup paths dynamically
ROOT = Path().resolve().parent if Path().resolve().name == "analysis" else Path().resolve()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from parser import load_fold
```

```python
def inspect(folder):
    X_train, y_train, X_test, y_test = load_fold(folder, 0)
    print("Train:", X_train.shape, " Test:", X_test.shape)
    print("\nClass counts:")
    print(pd.DataFrame({"train": y_train.value_counts(), "test": y_test.value_counts()}).fillna(0).astype(int))
    print("\nMajority class accuracy on test:", (y_test == y_train.mode()[0]).mean())
    print("\nMissing values in train (only columns with any):")
    missing = X_train.isna().sum()
    print(missing[missing > 0])
    return X_train, y_train, X_test, y_test
```

## Hypothyroid


```python
X_train, y_train, X_test, y_test = inspect("hypothyroid")
X_train.describe()
```
